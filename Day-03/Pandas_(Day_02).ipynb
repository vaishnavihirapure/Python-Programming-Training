```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y)
```

```python
#Bar Chart
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [30340, 33400, 28890, 32800, 42345, 27894, 42837, 67819, 89032, 84758, 19406, 26385]
plt.bar(months, sales, color='pink')
plt.title('Bar Chart')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
plt.barh(months, sales, color='pink')
plt.title('Monthly Sales Horizontal')
plt.xlabel('sales')
plt.ylabel('months')
plt.grid(True)
plt.show()
```

```python
#Scatter plot(comparing Height vs Weight)
height = [150,232,656,345]
weight = [23,54,32,64]
plt.scatter(height, weight, color='pink')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(True)
plt.show()
```

```python
# Pie Chart
brands = ['Apple', 'Samsung', 'Google', 'Microsoft']
shares = [27, 32, 21, 78]

plt.pie(shares, labels=brands, autopct='%1.1f%%',
        startangle=90, colors=['pink', 'purple', 'yellow', 'skyblue'])

plt.title('Market Share')
plt.axis('equal')
plt.show()
```

```python
#Mutiple Plot in one figure
x = [1,2,3,4]
y1 = [3,7,11,17]
y2 = [2,4,6,8]
plt.plot(x, y1, label='Square', color='pink')
plt.plot(x, y2, label='Linear', color='skyblue')
plt.legend()
plt.title('Multiple Lines')
plt.grid(True)
plt.show()
```

```python
#Subplots
fig, axs = plt.subplots(2, 2)
axs[0,0].plot(x,y1)
axs[0,0].set_title('Square')

axs[0,1].bar(months, sales)
axs[0,1].set_title('Sales')

axs[1,0].scatter(height, weight)
axs[1,0].set_title('Scatter')

axs[1,1].pie(shares, labels=brands, autopct='%1.1f%%')
axs[1,1].set_title('Pie')

plt.tight_layout()
plt.show()
```

```python
# Subplots

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# Line Chart
axs[0,0].plot(x, y1)
axs[0,0].set_title('Square')

# Bar Chart
axs[0,1].bar(months, sales)
axs[0,1].set_title('Sales')
axs[0,1].tick_params(axis='x', rotation=45)

# Scatter Plot
axs[1,0].scatter(height, weight)
axs[1,0].set_title('Scatter')

# Pie Chart
colors = ['#F8BBD0', '#E1BEE7', '#BBDEFB', '#FCE4EC']

axs[1,1].pie(
    shares,
    labels=brands,
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    labeldistance=1.15,
    pctdistance=0.65,
    textprops={'fontsize': 9}
)
axs[1,1].set_title('Pie')

plt.tight_layout()
plt.show()
```
