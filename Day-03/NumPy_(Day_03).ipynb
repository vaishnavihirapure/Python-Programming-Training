NumPy

```python
import numpy as np
students = np.array([
    [85,98,43],
    [43,63,34,],
    [73,93,86],
    [46,39,27],
    [53,34,48],
])
students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Marks :", avg_marks)
```

```python
avg_marks = np.mean(students, axis=0)
print("Average Marks :", avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper Student:", topper)
```

```python
failed_student = np.where(avg_marks<50)
print("Failed Student:", failed_student[0]+1)
```

```python
#Performance Report
for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"Student {i+1} : Avg = {avg:2f} | {status}")
```
