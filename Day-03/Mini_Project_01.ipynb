MINI PROJECT DATA ANALYSIS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Maths': [98, 89, 99, 87, 69],
    'Science': [79, 60, 78, 64, 68],
    'English': [54, 90, 99, 80, 90]

}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Maths", "Science", "English"]]
average_marks = np.mean(marks, axis=1)
df["Average Marks"] = average_marks
print(df)
```

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"
df["Grade"] = df["Average Marks"].apply(grade)
df
```

```python
plt.figure()
plt.bar(df["Name"], df["Average Marks"])
plt.xlabel("Name")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```

```python
# Average Marks per Subject

import matplotlib.pyplot as plt

df[['Maths', 'Science', 'English']].mean().plot(kind='bar')

plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks per Subject")

plt.show()
```
