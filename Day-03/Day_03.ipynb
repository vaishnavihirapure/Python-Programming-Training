```python
#Writing data to .csv file
import csv
with open('emp.csv',"w",newline='') as f:
    w=csv.writer(f)
    w.writerow(["END","ENAME","SSALARY","EADDR"])
    n=int(input("Enter number of employee: "))
    for i in range(n):
        eno=input("Enter employee number: ")
        ename=input("Enter employee name: ")
        esal=input("Enter employee salary: ")
        eaddr=input("Enter employee address: ")
        w.writerow([eno,ename,esal,eaddr])
print("Data inserted successfully")
```

```python
#Reading Data from .csv file
import csv
f=open("emp.csv",'r')
r=csv.reader(f)
data=list(r)
print(data)                                 #print in lidt format
for line in data:
    for word in line:
        print(word,"\t \t",end='')
    print()
```

```python
#ZIP Operation
from zipfile import *
f=ZipFile("myzips.zip",'w',ZIP_DEFLATED)
f.write("emp.csv")
f.write("mochii.jpg")
f.write("newmochii.jpg")
f.close()
print("myzips.zip is created")
```
